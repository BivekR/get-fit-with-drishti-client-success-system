```python
# Import the main libraries we need for data loading and cleaning
import pandas as pd
import numpy as np

# Show all columns when displaying the dataset
pd.set_option("display.max_columns", None)

# Set the path to the raw CSV file
# The notebook is inside 03_python_notebooks
# ../ means go one folder back to the main project folder
raw_file_path = "../01_raw_data/Get_Fit_with_Drishti_Business_Data.csv"

# Load the raw CSV file into a pandas DataFrame
df = pd.read_csv(raw_file_path)

# Print the number of rows and columns
print("Dataset shape:")
print(df.shape)

# Print all column names
print("\nColumn names:")
print(df.columns.tolist())

# Print the data type of each column
print("\nData types:")
print(df.dtypes)

# Show the first 3 rows of the dataset
print("\nFirst 3 rows:")
display(df.head(3))

# Print the number of missing values in each column
print("\nNull counts per column:")
print(df.isnull().sum())
```

```python
# Privacy protection
# Drop Client_Name because it contains real client names
# Keep Client_ID because it is anonymous and still useful for analysis

df = df.drop(columns=["Client_Name"])

# Confirm that Client_Name was removed
print("Client_Name exists:", "Client_Name" in df.columns)

# Check the new dataset shape
print("Shape after dropping Client_Name:", df.shape)
```

```python
# Convert date columns
# These date columns are currently stored as text in DD/MM/YYYY format
# We convert them into real datetime columns so Python can analyze dates correctly

df["Payment_Date"] = pd.to_datetime(df["Payment_Date"], dayfirst=True)
df["Program_Start_Date"] = pd.to_datetime(df["Program_Start_Date"], dayfirst=True)
df["Program_End_Date"] = pd.to_datetime(df["Program_End_Date"], dayfirst=True)

# Confirm that the three columns are now datetime type
print(df[["Payment_Date", "Program_Start_Date", "Program_End_Date"]].dtypes)

# Preview the converted date columns
df[["Payment_Date", "Program_Start_Date", "Program_End_Date"]].head()
```

```python
# Create Program_Duration_Days
# This calculates the number of days between Program_Start_Date and Program_End_Date

df["Program_Duration_Days"] = (
    df["Program_End_Date"] - df["Program_Start_Date"]
).dt.days

# Preview the new duration column with the original date columns
df[["Program_Start_Date", "Program_End_Date", "Program_Duration_Days"]].head()
```

```python
#  Clean revenue columns
# Drop Revenue and Revenue_Currency because Revenue contains mixed currencies
# Keep Revenue_USD because it is already normalized into USD

df = df.drop(columns=["Revenue", "Revenue_Currency"])

# Confirm only Revenue_USD remains
print("Revenue-related columns left:")
print([col for col in df.columns if "Revenue" in col])

# Check the new dataset shape
print("Shape after dropping Revenue and Revenue_Currency:", df.shape)
```

```python
# Clean Days_To_First_Engagement
# This column has numbers mixed with the text value "Never"
# We replace "Never" with 999 so the whole column can become numeric

df["Days_To_First_Engagement"] = df["Days_To_First_Engagement"].replace("Never", 999)

# Convert the column from text/object type to integer
df["Days_To_First_Engagement"] = df["Days_To_First_Engagement"].astype(int)

# Create a new boolean column to show whether the client ever engaged
# True means the client engaged at least once
# False means the client never engaged
df["Ever_Engaged"] = df["Days_To_First_Engagement"] != 999

# Confirm the cleaned column type
print("Days_To_First_Engagement data type:")
print(df["Days_To_First_Engagement"].dtype)

# Check how many clients engaged vs never engaged
print("\nEver_Engaged counts:")
print(df["Ever_Engaged"].value_counts())

# Preview the two columns together
df[["Days_To_First_Engagement", "Ever_Engaged"]].head(10)
```

```python
# Clean referral data
# Missing values in Referred_By mean the client was not referred by anyone
# We fill those blanks with "None"

df["Referred_By"] = df["Referred_By"].fillna("None")

# Create a new boolean column to show whether the client was referred
# True means the client has a referrer
# False means the client was not referred
df["Was_Referred"] = df["Referred_By"] != "None"

# Confirm there are no missing values left in Referred_By
print("Missing Referred_By values:")
print(df["Referred_By"].isnull().sum())

# Count referred vs not referred clients
print("\nWas_Referred counts:")
print(df["Was_Referred"].value_counts())

# Preview the referral columns
df[["Client_ID", "Referred_By", "Was_Referred"]].head(10)
```

```python
#Clean Health_Condition
# Missing values in Health_Condition mean the client has no recorded health condition
# We fill those blanks with "None"

df["Health_Condition"] = df["Health_Condition"].fillna("None")

# Confirm there are no missing values left in Health_Condition
print("Missing Health_Condition values:")
print(df["Health_Condition"].isnull().sum())

# Check the count of each health condition category
print("\nHealth_Condition counts:")
print(df["Health_Condition"].value_counts())
```

```python
# Clean Exercise_History
# Exercise_History has some missing values
# We will fill them using the mode, which is the most common value in the column

exercise_mode = df["Exercise_History"].mode()[0]

# Fill missing Exercise_History values with the most common category
df["Exercise_History"] = df["Exercise_History"].fillna(exercise_mode)

# Confirm which value was used
print("Mode used for Exercise_History:")
print(exercise_mode)

# Confirm there are no missing values left in Exercise_History
print("\nMissing Exercise_History values:")
print(df["Exercise_History"].isnull().sum())

# Check the updated Exercise_History counts
print("\nExercise_History counts:")
print(df["Exercise_History"].value_counts())
```

```python
# Clean dropout-related columns
# These columns are blank for clients who never dropped out
# That is expected, so we fill them with meaningful values

df["Dropout_Day"] = df["Dropout_Day"].fillna(0)
df["Dropout_Reason"] = df["Dropout_Reason"].fillna("N/A")
df["Returned_After_Drop"] = df["Returned_After_Drop"].fillna("N/A")
df["Weeks_Before_Return"] = df["Weeks_Before_Return"].fillna(0)
df["Return_Trigger"] = df["Return_Trigger"].fillna("N/A")

# List the dropout-related columns so we can check them together
dropout_cols = [
    "Dropout_Day",
    "Dropout_Reason",
    "Returned_After_Drop",
    "Weeks_Before_Return",
    "Return_Trigger"
]

# Confirm there are no missing values left in these columns
print("Missing values in dropout-related columns:")
print(df[dropout_cols].isnull().sum())

# Preview the cleaned dropout-related columns
df[dropout_cols].head(10)
```

```python
# Drop redundant Taste_Preference column
# Taste_Preference duplicates the same information as Diet_Type
# We remove it to keep the dataset cleaner and avoid repeated analysis

df = df.drop(columns=["Taste_Preference"])

# Confirm that Taste_Preference was removed
print("Taste_Preference exists:", "Taste_Preference" in df.columns)

# Check the new dataset shape
print("Shape after dropping Taste_Preference:", df.shape)
```

```python
#Verify final cleaned dataset
# This step checks whether the cleaned dataset is ready to save and use for SQL, EDA, Tableau, and reporting

# Print final dataset shape
print("Final dataset shape:")
print(df.shape)

# Confirm date column data types
print("\nDate column data types:")
print(df[["Payment_Date", "Program_Start_Date", "Program_End_Date"]].dtypes)

# Confirm Days_To_First_Engagement is now integer
print("\nDays_To_First_Engagement data type:")
print(df["Days_To_First_Engagement"].dtype)

# Confirm engineered columns exist
print("\nEngineered columns check:")
engineered_columns = ["Program_Duration_Days", "Ever_Engaged", "Was_Referred"]

for col in engineered_columns:
    print(col, "exists:", col in df.columns)

# Check all remaining null values
print("\nRemaining null counts:")
null_counts = df.isnull().sum()
print(null_counts[null_counts > 0])

# Final success/warning message
if null_counts.sum() == 0:
    print("\nSuccess: There are zero remaining null values.")
else:
    print("\nWarning: Some null values still exist.")
```

```python
# Save the cleaned dataset
# This saves the cleaned DataFrame as a new CSV file inside the 02_cleaned_data folder

clean_file_path = "../02_cleaned_data/Drishti_Clean_Data.csv"

# Save the cleaned dataset
# index=False prevents pandas from adding an extra unnecessary index column
df.to_csv(clean_file_path, index=False)

# Confirm the file was saved
print("Cleaned dataset saved successfully:")
print(clean_file_path)
```
