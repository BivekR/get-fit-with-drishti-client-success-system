```python
# SQL Query 1:Goal Type Breakdown
# This query counts clients by goal type and calculates percentage of total clients

query_1 = """
SELECT 
    Goal_Type,
    COUNT(*) AS client_count,
    ROUND(COUNT(*) * 100.0 / (SELECT COUNT(*) FROM clients), 2) AS percentage_of_total
FROM clients
GROUP BY Goal_Type
ORDER BY client_count DESC;
"""

# Run the SQL query and store the result in a DataFrame
query_1_result = pd.read_sql_query(query_1, conn)

# Display the result
query_1_result
```

```python
# Save into the SQL file
# "a" means append, so it adds this query without deleting the existing file content

with open(sql_file_path, "a") as file:
    file.write("""
-- Query 1: Goal Type Breakdown
-- Business Question:
-- What are the main client goals, and what percentage of total clients does each goal represent?

SELECT 
    Goal_Type,
    COUNT(*) AS client_count,
    ROUND(COUNT(*) * 100.0 / (SELECT COUNT(*) FROM clients), 2) AS percentage_of_total
FROM clients
GROUP BY Goal_Type
ORDER BY client_count DESC;

""")

# Print confirmation message
print("Query 1 saved to SQL file successfully.")
```

```python
# SQL Query 2:Lead Source Completion Rate
# This query finds the completion rate for each lead source

query_2 = """
SELECT
    Lead_Source,
    COUNT(*) AS total_clients,
    SUM(CASE WHEN Status = 'Completed' THEN 1 ELSE 0 END) AS completed_clients,
    ROUND(
        SUM(CASE WHEN Status = 'Completed' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS completion_rate_pct
FROM clients
GROUP BY Lead_Source
ORDER BY completion_rate_pct DESC;
"""

# Run the SQL query and store the result in a DataFrame
query_2_result = pd.read_sql_query(query_2, conn)

# Display the result
query_2_result
```

```python
#  SQL Query 3:Average Check-in Rate by Goal Type and Motivation Level
# This query shows average client engagement across goal type and starting motivation level

query_3 = """
SELECT
    Goal_Type,
    Motivation_Level_Start,
    COUNT(*) AS total_clients,
    ROUND(AVG(Checkin_Rate_Pct), 2) AS avg_checkin_rate_pct
FROM clients
GROUP BY Goal_Type, Motivation_Level_Start
ORDER BY Goal_Type, avg_checkin_rate_pct DESC;
"""

# Run the SQL query and store the result in a DataFrame
query_3_result = pd.read_sql_query(query_3, conn)

# Display the result
query_3_result
```

```python
# SQL Query 4: Goal Achievement Rate by Diet Type
# This query compares success rate across different diet types

query_4 = """
SELECT
    Diet_Type,
    COUNT(*) AS total_clients,
    SUM(CASE WHEN Goal_Achieved = 'Yes' THEN 1 ELSE 0 END) AS goal_achieved_clients,
    ROUND(
        SUM(CASE WHEN Goal_Achieved = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS goal_achievement_rate_pct
FROM clients
GROUP BY Diet_Type
ORDER BY goal_achievement_rate_pct DESC;
"""

# Run the SQL query and store the result in a DataFrame
query_4_result = pd.read_sql_query(query_4, conn)

# Display the result
query_4_result
```

```python
# SQL Query 5:Country Breakdown
# This query compares client count, completion rate, and goal achievement rate by country

query_5 = """
SELECT
    Country,
    COUNT(*) AS total_clients,
    SUM(CASE WHEN Status = 'Completed' THEN 1 ELSE 0 END) AS completed_clients,
    ROUND(
        SUM(CASE WHEN Status = 'Completed' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS completion_rate_pct,
    SUM(CASE WHEN Goal_Achieved = 'Yes' THEN 1 ELSE 0 END) AS goal_achieved_clients,
    ROUND(
        SUM(CASE WHEN Goal_Achieved = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS goal_achievement_rate_pct
FROM clients
GROUP BY Country
ORDER BY total_clients DESC;
"""

# Run the SQL query and store the result in a DataFrame
query_5_result = pd.read_sql_query(query_5, conn)

# Display the result
query_5_result
```

```python
# SQL Query 6:Monthly Client Acquisition Trend
# This query counts how many new clients joined each month using Payment_Date

query_6 = """
SELECT
    strftime('%Y-%m', Payment_Date) AS payment_month,
    COUNT(*) AS new_clients
FROM clients
GROUP BY payment_month
ORDER BY payment_month;
"""

# Run the SQL query and store the result in a DataFrame
query_6_result = pd.read_sql_query(query_6, conn)

# Display the result
query_6_result
```

```python
# SQL Query 7:Week 3 Activity and Completion
# This query checks week 3 activity and whether inactive clients eventually completed

query_7 = """
SELECT
    Week3_Still_Active,
    COUNT(*) AS total_clients,
    ROUND(COUNT(*) * 100.0 / (SELECT COUNT(*) FROM clients), 2) AS percentage_of_total,
    SUM(CASE WHEN Status = 'Completed' THEN 1 ELSE 0 END) AS completed_clients,
    ROUND(
        SUM(CASE WHEN Status = 'Completed' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS completion_rate_pct
FROM clients
GROUP BY Week3_Still_Active
ORDER BY Week3_Still_Active DESC;
"""

# Run the SQL query and store the result in a DataFrame
query_7_result = pd.read_sql_query(query_7, conn)

# Display the result
query_7_result
```

```python
# SQL Query 8:Second Round Analysis
# This query checks what percentage of completed clients signed up again, grouped by goal type

query_8 = """
SELECT
    Goal_Type,
    COUNT(*) AS completed_clients,
    SUM(CASE WHEN Second_Round = 'Yes' THEN 1 ELSE 0 END) AS second_round_clients,
    ROUND(
        SUM(CASE WHEN Second_Round = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS second_round_rate_pct
FROM clients
WHERE Status = 'Completed'
GROUP BY Goal_Type
ORDER BY second_round_rate_pct DESC;
"""

# Run the SQL query and store the result in a DataFrame
query_8_result = pd.read_sql_query(query_8, conn)

# Display the result
query_8_result
```

```python
# SQL Query 9:Ghost Client Profile
# This query summarizes clients who had 0% check-in rate

query_9 = """
SELECT
    Goal_Type,
    Lead_Source,
    Country,
    COUNT(*) AS ghost_clients,
    ROUND(AVG(Age), 2) AS avg_age
FROM clients
WHERE Checkin_Rate_Pct = 0
GROUP BY Goal_Type, Lead_Source, Country
ORDER BY ghost_clients DESC;
"""

# Run the SQL query and store the result in a DataFrame
query_9_result = pd.read_sql_query(query_9, conn)

# Display the result
query_9_result
```

```python
#  SQL Query 10:Payment Speed by Lead Source
# This query compares how fast clients from each lead source convert from first contact to payment

query_10 = """
SELECT
    Lead_Source,
    COUNT(*) AS total_clients,
    ROUND(AVG(Payment_Speed_Days), 2) AS avg_payment_speed_days,
    MIN(Payment_Speed_Days) AS fastest_payment_days,
    MAX(Payment_Speed_Days) AS slowest_payment_days
FROM clients
GROUP BY Lead_Source
ORDER BY avg_payment_speed_days ASC;
"""

# Run the SQL query and store the result in a DataFrame
query_10_result = pd.read_sql_query(query_10, conn)

# Display the result
query_10_result
```

```python
# SQL Query 12: Win-back Analysis
# This query analyzes dropout clients and checks how many returned after dropping out

query_12 = """
SELECT
    Dropout_Reason,
    COUNT(*) AS dropout_clients,
    SUM(CASE WHEN Returned_After_Drop = 'Yes' THEN 1 ELSE 0 END) AS returned_clients,
    ROUND(
        SUM(CASE WHEN Returned_After_Drop = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS return_rate_pct,
    ROUND(AVG(Weeks_Before_Return), 2) AS avg_weeks_before_return
FROM clients
WHERE Dropout_Reason IS NOT NULL
GROUP BY Dropout_Reason
ORDER BY return_rate_pct DESC;
"""

# Run the SQL query and store the result in a DataFrame
query_12_result = pd.read_sql_query(query_12, conn)

# Display the result
query_12_result
```

```python
# SQL Query 13: Referral Value
# This query compares referred clients vs non-referred clients

query_13 = """
SELECT
    CASE
        WHEN Referred_By IS NULL THEN 'Non-Referred Client'
        ELSE 'Referred Client'
    END AS referral_group,
    COUNT(*) AS total_clients,
    ROUND(AVG(Revenue_USD), 2) AS avg_revenue_usd,
    SUM(CASE WHEN Status = 'Completed' THEN 1 ELSE 0 END) AS completed_clients,
    ROUND(
        SUM(CASE WHEN Status = 'Completed' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS completion_rate_pct,
    SUM(CASE WHEN Second_Round = 'Yes' THEN 1 ELSE 0 END) AS second_round_clients,
    ROUND(
        SUM(CASE WHEN Second_Round = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS second_round_rate_pct
FROM clients
GROUP BY referral_group
ORDER BY completion_rate_pct DESC;
"""

# Run the SQL query and store the result in a DataFrame
query_13_result = pd.read_sql_query(query_13, conn)

# Display the result
query_13_result
```

```python
# Save all 13 finalized SQL queries into one clean .sql file
# This will overwrite the existing file and create the final organized SQL query file

all_sql_queries = """
-- ============================================================
-- Get Fit with Drishti
-- Phase 2: SQL Business Analysis Queries
-- Database: nutricoach.db
-- Table: clients
-- ============================================================


-- Query 1: Goal Type Breakdown
-- Business Question:
-- What are the main client goals, and what percentage of total clients does each goal represent?

SELECT 
    Goal_Type,
    COUNT(*) AS client_count,
    ROUND(COUNT(*) * 100.0 / (SELECT COUNT(*) FROM clients), 2) AS percentage_of_total
FROM clients
GROUP BY Goal_Type
ORDER BY client_count DESC;



-- Query 2: Lead Source Completion Rate
-- Business Question:
-- Which lead source brings clients with the highest completion rate?

SELECT
    Lead_Source,
    COUNT(*) AS total_clients,
    SUM(CASE WHEN Status = 'Completed' THEN 1 ELSE 0 END) AS completed_clients,
    ROUND(
        SUM(CASE WHEN Status = 'Completed' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS completion_rate_pct
FROM clients
GROUP BY Lead_Source
ORDER BY completion_rate_pct DESC;



-- Query 3: Average Check-in Rate by Goal Type and Motivation Level
-- Business Question:
-- Which goal and motivation groups have the highest or lowest check-in engagement?

SELECT
    Goal_Type,
    Motivation_Level_Start,
    COUNT(*) AS total_clients,
    ROUND(AVG(Checkin_Rate_Pct), 2) AS avg_checkin_rate_pct
FROM clients
GROUP BY Goal_Type, Motivation_Level_Start
ORDER BY Goal_Type, avg_checkin_rate_pct DESC;



-- Query 4: Goal Achievement Rate by Diet Type
-- Business Question:
-- Does diet type appear to affect client success?

SELECT
    Diet_Type,
    COUNT(*) AS total_clients,
    SUM(CASE WHEN Goal_Achieved = 'Yes' THEN 1 ELSE 0 END) AS goal_achieved_clients,
    ROUND(
        SUM(CASE WHEN Goal_Achieved = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS goal_achievement_rate_pct
FROM clients
GROUP BY Diet_Type
ORDER BY goal_achievement_rate_pct DESC;



-- Query 5: Country Breakdown
-- Business Question:
-- Which countries bring the most clients, and how successful are clients from each country?

SELECT
    Country,
    COUNT(*) AS total_clients,
    SUM(CASE WHEN Status = 'Completed' THEN 1 ELSE 0 END) AS completed_clients,
    ROUND(
        SUM(CASE WHEN Status = 'Completed' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS completion_rate_pct,
    SUM(CASE WHEN Goal_Achieved = 'Yes' THEN 1 ELSE 0 END) AS goal_achieved_clients,
    ROUND(
        SUM(CASE WHEN Goal_Achieved = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS goal_achievement_rate_pct
FROM clients
GROUP BY Country
ORDER BY total_clients DESC;



-- Query 6: Monthly Client Acquisition Trend
-- Business Question:
-- How many new clients joined each month?

SELECT
    strftime('%Y-%m', Payment_Date) AS payment_month,
    COUNT(*) AS new_clients
FROM clients
GROUP BY payment_month
ORDER BY payment_month;



-- Query 7: Week 3 Activity and Completion
-- Business Question:
-- Is week 3 activity a strong early warning signal for program completion?

SELECT
    Week3_Still_Active,
    COUNT(*) AS total_clients,
    ROUND(COUNT(*) * 100.0 / (SELECT COUNT(*) FROM clients), 2) AS percentage_of_total,
    SUM(CASE WHEN Status = 'Completed' THEN 1 ELSE 0 END) AS completed_clients,
    ROUND(
        SUM(CASE WHEN Status = 'Completed' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS completion_rate_pct
FROM clients
GROUP BY Week3_Still_Active
ORDER BY Week3_Still_Active DESC;



-- Query 8: Second Round Analysis
-- Business Question:
-- Which completed client goal group is most likely to sign up again?

SELECT
    Goal_Type,
    COUNT(*) AS completed_clients,
    SUM(CASE WHEN Second_Round = 'Yes' THEN 1 ELSE 0 END) AS second_round_clients,
    ROUND(
        SUM(CASE WHEN Second_Round = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS second_round_rate_pct
FROM clients
WHERE Status = 'Completed'
GROUP BY Goal_Type
ORDER BY second_round_rate_pct DESC;



-- Query 9: Ghost Client Profile
-- Business Question:
-- Who are the clients most likely to disappear without checking in?

SELECT
    Goal_Type,
    Lead_Source,
    Country,
    COUNT(*) AS ghost_clients,
    ROUND(AVG(Age), 2) AS avg_age
FROM clients
WHERE Checkin_Rate_Pct = 0
GROUP BY Goal_Type, Lead_Source, Country
ORDER BY ghost_clients DESC;



-- Query 10: Payment Speed by Lead Source
-- Business Question:
-- Which lead source converts clients from first contact to payment the fastest?

SELECT
    Lead_Source,
    COUNT(*) AS total_clients,
    ROUND(AVG(Payment_Speed_Days), 2) AS avg_payment_speed_days,
    MIN(Payment_Speed_Days) AS fastest_payment_days,
    MAX(Payment_Speed_Days) AS slowest_payment_days
FROM clients
GROUP BY Lead_Source
ORDER BY avg_payment_speed_days ASC;



-- Query 11: Health Condition vs Healthy Clients
-- Business Question:
-- Do clients with health conditions perform differently from clients without health conditions?

SELECT
    CASE 
        WHEN Health_Condition IS NULL THEN 'No Health Condition'
        ELSE 'Has Health Condition'
    END AS health_group,
    COUNT(*) AS total_clients,
    SUM(CASE WHEN Status = 'Completed' THEN 1 ELSE 0 END) AS completed_clients,
    ROUND(
        SUM(CASE WHEN Status = 'Completed' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS completion_rate_pct,
    SUM(CASE WHEN Goal_Achieved = 'Yes' THEN 1 ELSE 0 END) AS goal_achieved_clients,
    ROUND(
        SUM(CASE WHEN Goal_Achieved = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS goal_achievement_rate_pct
FROM clients
GROUP BY health_group
ORDER BY total_clients DESC;



-- Query 12: Win-back Analysis
-- Business Question:
-- Which dropout reasons are most likely to turn into successful win-backs?

SELECT
    Dropout_Reason,
    COUNT(*) AS dropout_clients,
    SUM(CASE WHEN Returned_After_Drop = 'Yes' THEN 1 ELSE 0 END) AS returned_clients,
    ROUND(
        SUM(CASE WHEN Returned_After_Drop = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS return_rate_pct,
    ROUND(AVG(Weeks_Before_Return), 2) AS avg_weeks_before_return
FROM clients
WHERE Dropout_Reason IS NOT NULL
GROUP BY Dropout_Reason
ORDER BY return_rate_pct DESC;



-- Query 13: Referral Value
-- Business Question:
-- Are referred clients more valuable than non-referred clients?

SELECT
    CASE
        WHEN Referred_By IS NULL THEN 'Non-Referred Client'
        ELSE 'Referred Client'
    END AS referral_group,
    COUNT(*) AS total_clients,
    ROUND(AVG(Revenue_USD), 2) AS avg_revenue_usd,
    SUM(CASE WHEN Status = 'Completed' THEN 1 ELSE 0 END) AS completed_clients,
    ROUND(
        SUM(CASE WHEN Status = 'Completed' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS completion_rate_pct,
    SUM(CASE WHEN Second_Round = 'Yes' THEN 1 ELSE 0 END) AS second_round_clients,
    ROUND(
        SUM(CASE WHEN Second_Round = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS second_round_rate_pct
FROM clients
GROUP BY referral_group
ORDER BY completion_rate_pct DESC;
"""

# Write all final SQL queries into the .sql file
with open(sql_file_path, "w") as file:
    file.write(all_sql_queries)

# Print confirmation message
print("All 13 SQL queries saved successfully.")
print("File location:", sql_file_path)
```
